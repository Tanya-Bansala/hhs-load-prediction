```python
df = pd.read_csv("../data/HHS_Unaccompanied_Alien_Children_Program.csv")

df = df.dropna(how='all')

df.head()
```

```python
df['Children in HHS Care'].head(10)
```

```python
df['Children in HHS Care'] = df['Children in HHS Care'].str.replace(',', '')

df['Children in HHS Care'] = pd.to_numeric(df['Children in HHS Care'])

df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df.info()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))
plt.plot(df['Date'], df['Children in HHS Care'])
plt.title("HHS Care Load Over Time")
plt.xlabel("Date")
plt.ylabel("Children in HHS Care")
plt.show()
```

```python
plt.figure(figsize=(12,5))

# Intake (CBP custody me aane wale children)
plt.plot(df['Date'], df['Children apprehended and placed in CBP custody*'], 
         label='Children entering system', alpha=0.7)

# Discharge (HHS se safe placement wale children)
plt.plot(df['Date'], df['Children discharged from HHS Care'], 
         label='Children exiting system', alpha=0.7)

plt.title("Children Entering vs Exiting Care System Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Children")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,5))

# 7-day rolling average (smooth trend)
df['HHS_Rolling'] = df['Children in HHS Care'].rolling(window=7).mean()

plt.plot(df['Date'], df['Children in HHS Care'], 
         alpha=0.3, label='Actual')

plt.plot(df['Date'], df['HHS_Rolling'], 
         color='red', label='7-Day Rolling Average')

plt.title("HHS Care Load Trend (Smoothed)")
plt.xlabel("Date")
plt.ylabel("Children in HHS Care")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,5))

# Net pressure = intake - discharge
df['Net_Pressure'] = df['Children apprehended and placed in CBP custody*'] - df['Children discharged from HHS Care']

plt.plot(df['Date'], df['Net_Pressure'], color='purple')

plt.title("Net System Pressure (Intake - Discharge)")
plt.xlabel("Date")
plt.ylabel("Net Pressure")
plt.axhline(0, color='black', linestyle='--')
plt.show()
```

```python
# Previous day's HHS care load
df['HHS_Care_Previous_Day'] = df['Children in HHS Care'].shift(1)

# Previous 7 days HHS care load
df['HHS_Care_7_Days_Back'] = df['Children in HHS Care'].shift(7)

# Display new columns
df[['Children in HHS Care',
    'HHS_Care_Previous_Day',
    'HHS_Care_7_Days_Back']].head(10)
```

```python
# 7-day rolling average
df['HHS_Care_7Day_Average'] = (
    df['Children in HHS Care']
    .rolling(window=7)
    .mean()
)

# 14-day rolling average
df['HHS_Care_14Day_Average'] = (
    df['Children in HHS Care']
    .rolling(window=14)
    .mean()
)

# Display results
df[['Children in HHS Care',
    'HHS_Care_7Day_Average',
    'HHS_Care_14Day_Average']].head(20)
```

```python
# Remove rows with missing values created by lag/rolling features
df = df.dropna()

# Select target column
target_column = 'Children in HHS Care'

# Create train-test split (80% train, 20% test)
split_index = int(len(df) * 0.8)

train_data = df.iloc[:split_index]
test_data = df.iloc[split_index:]

# Display dataset sizes
print("Training Data Size:", len(train_data))
print("Testing Data Size:", len(test_data))
```

```python
# Create naive predictions using previous day's value
test_data['Naive_Prediction'] = test_data['Children in HHS Care'].shift(1)

# Remove first missing prediction
test_data = test_data.dropna()

# Display actual vs predicted values
test_data[['Children in HHS Care', 'Naive_Prediction']].head()
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error

# Actual values
actual_values = test_data['Children in HHS Care']

# Predicted values
predicted_values = test_data['Naive_Prediction']

# Calculate MAE
mae = mean_absolute_error(actual_values, predicted_values)

# Calculate RMSE
rmse = mean_squared_error(actual_values, predicted_values) ** 0.5

# Calculate MAPE
mape = mean_absolute_percentage_error(actual_values, predicted_values)

# Display results
print("MAE:", mae)
print("RMSE:", rmse)
print("MAPE:", mape)
```

```python
# Set chart size
plt.figure(figsize=(12,5))

# Plot actual values
plt.plot(
    test_data['Date'],
    test_data['Children in HHS Care'],
    label='Actual Values'
)

# Plot predicted values
plt.plot(
    test_data['Date'],
    test_data['Naive_Prediction'],
    label='Predicted Values'
)

# Add chart title and labels
plt.title("Actual vs Predicted HHS Care Load")
plt.xlabel("Date")
plt.ylabel("Children in HHS Care")

# Show legend
plt.legend()

# Display chart
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

# Select input features
feature_columns = [
    'HHS_Care_Previous_Day',
    'HHS_Care_7_Days_Back',
    'HHS_Care_7Day_Average',
    'HHS_Care_14Day_Average',
    'Net_Pressure'
]

# Training features and target
X_train = train_data[feature_columns]
y_train = train_data['Children in HHS Care']

# Testing features and target
X_test = test_data[feature_columns]
y_test = test_data['Children in HHS Care']

# Create Random Forest model
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Train the model
rf_model.fit(X_train, y_train)

# Generate predictions
rf_predictions = rf_model.predict(X_test)

# Display first predictions
print(rf_predictions[:5])
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error

# Calculate MAE
rf_mae = mean_absolute_error(y_test, rf_predictions)

# Calculate RMSE
rf_rmse = mean_squared_error(y_test, rf_predictions) ** 0.5

# Calculate MAPE
rf_mape = mean_absolute_percentage_error(y_test, rf_predictions)

# Print results
print("Random Forest MAE:", rf_mae)
print("Random Forest RMSE:", rf_rmse)
print("Random Forest MAPE:", rf_mape)
```

```python
# Set chart size
plt.figure(figsize=(12,5))

# Plot actual values
plt.plot(
    test_data['Date'],
    y_test,
    label='Actual Values'
)

# Plot Random Forest predictions
plt.plot(
    test_data['Date'],
    rf_predictions,
    label='Random Forest Predictions'
)

# Add title and labels
plt.title("Random Forest Forecast vs Actual Values")
plt.xlabel("Date")
plt.ylabel("Children in HHS Care")

# Display legend
plt.legend()

# Show chart
plt.show()
```

```python

# Select last 30 rows from dataset
future_forecast_data = df.tail(30)

# Generate predictions using Random Forest model
future_predictions = rf_model.predict(
    future_forecast_data[feature_columns]
)

# Create forecast dataframe
forecast_results = future_forecast_data[['Date']].copy()

forecast_results['Predicted_HHS_Care_Load'] = future_predictions

# Display forecast results
forecast_results.head()
```
